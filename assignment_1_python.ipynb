#10.print("\nQuestion 10: Calculate the sum of first 20 odd numbers")

```python
total = 0
count = 0
for i in range(1, 41):
    if i % 2 != 0:
        total = total + i
        count = count + 1
    if count == 20:
        break
print(total)
```
