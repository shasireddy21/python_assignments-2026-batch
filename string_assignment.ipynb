```python
#41: Check if string contains balanced parentheses: "((()))"
text = "((()))"
count = 0
balanced = True

for char in text:
    if char == "(":
        count += 1
    elif char == ")":
        count -= 1
    if count < 0:
        balanced = False
        break

if count != 0:
    balanced = False

print(balanced)
```

```python
#43: Find longest common substring

text1 = "programming"
text2 = "grammar"
longest = ""

for i in range(len(text1)):
    for j in range(i + 1, len(text1) + 1):
        sub = text1[i:j]

        if sub in text2 and len(sub) > len(longest):
            longest = sub

print(longest)
```

```python
#47: Check valid IPv4 address

ip = "192.168.1.1"
parts = ip.split(".")
valid = len(parts) == 4

if valid:
    for part in parts:
        if not part.isdigit() or not 0 <= int(part) <= 255:
            valid = False
            break
print(valid)
```

```python
# 49: Convert "hello world" to ROT13 encoding

text = "hello world"
result = ""

for char in text:
    if char == " ":
        result += " "
    else:
        result += chr((ord(char) - ord("a") + 13) % 26 + ord("a"))

print("ROT13:", result)
```

```python
# 50: Check if string is a valid credit card number: "4532015112830366"
text = "4532015112830366"
total = 0

for i in range(len(text)):
    digit = int(text[i])

    if i % 2 == 0:
        digit = digit * 2

        if digit > 9:
            digit = digit - 9
    total += digit
if total % 10 == 0:
    print("Valid ")
else:
    print("Invalid c")
```
